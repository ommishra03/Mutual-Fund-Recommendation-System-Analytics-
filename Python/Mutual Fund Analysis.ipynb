## Data Cleaning 

```python
print(df[['returns_3yr', 'returns_5yr']].dtypes)
```

```python
df['returns_3yr'] = pd.to_numeric(df['returns_3yr'], errors='coerce')
df['returns_5yr'] = pd.to_numeric(df['returns_5yr'], errors='coerce')
```

```python
mean_returns_3yr = df['returns_3yr'].mean()
mean_returns_5yr = df['returns_5yr'].mean()

df['returns_3yr'] = df['returns_3yr'].fillna(mean_returns_3yr)
df['returns_5yr'] = df['returns_5yr'].fillna(mean_returns_5yr)
```

```python
print(df.isnull().sum())
print(df[['returns_3yr', 'returns_5yr']].dtypes)
```

```python
# Replace missing values in returns_3yr and returns_5yr
mean_returns_3yr = df['returns_3yr'].mean()
mean_returns_5yr = df['returns_5yr'].mean()
df['returns_3yr'].fillna(mean_returns_3yr, inplace=True)
df['returns_5yr'].fillna(mean_returns_5yr, inplace=True)
print(df.isnull().sum())
```

## Data Normalization

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

columns_to_normalize = [
    'expense_ratio',
    'returns_1yr',
    'returns_3yr',
    'returns_5yr',
    'sharpe',
    'sortino_ratio',
    'alpha',
    'beta'
]

# Convert invalid entries to NaN, then numeric
df[columns_to_normalize] = (
    df[columns_to_normalize]
    .replace('-', pd.NA)
    .apply(pd.to_numeric, errors='coerce')
)

scaler = MinMaxScaler()

df_normalized = pd.DataFrame(
    scaler.fit_transform(df[columns_to_normalize]),
    columns=columns_to_normalize,
    index=df.index
)

# Invert metrics where lower is better
df_normalized['expense_ratio'] = 1 - df_normalized['expense_ratio']
df_normalized['beta'] = 1 - df_normalized['beta']
```

```python
missing_cols = set(columns_to_normalize) - set(df.columns)
print("Missing columns:", missing_cols)
```

## Composite Scoring

```python
weights = {
    'expense_ratio': 0.2,
    'returns_1yr': 0.15,
    'returns_3yr': 0.15,
    'returns_5yr': 0.15,
    'sharpe': 0.1,
    'sortino_ratio': 0.1,
    'alpha': 0.1,
    'beta': 0.05
}

df_normalized['composite_score'] = sum(
    df_normalized[col] * weight for col, weight in weights.items()
)
df['composite_score'] = df_normalized['composite_score']
```

## Ranking Fund

```python
df['rank'] = df['composite_score'].rank(ascending=False)
df_sorted = df.sort_values(by='rank')
```

## Exporting Result

```python
output_path = "../Excel/mutual_funds_.xlsx"

df_sorted.to_excel(output_path, index=False)

print(f"Exported full cleaned dataset ({df_sorted.shape[0]} rows) to '{output_path}'")
```
